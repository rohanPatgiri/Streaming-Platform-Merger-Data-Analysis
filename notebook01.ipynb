## 2. Analysis Phase

### 1. Content Library Analysis

```python
jotstarContents.head()
```

```python
LioCinemaContents.head()
```

```python
jotstarContents['platform'] = "Jotstar"
LioCinemaContents['platform'] = "LioCinema"
```

```python
LioCinemaContents.head()
```

```python
jotstarContents.head()
```

```python
# Combine both the data frames
combined_contents = pd.concat([LioCinemaContents, jotstarContents], ignore_index = "True")
```

```python
combined_contents.tail()
```

```python
combined_contents.shape
```

```python
# Content Type Analysis
combined_contents['content_type']. unique()
```

- 1.1 Group Content by Content_Type (Movie Or Series) for each platform

```python
content_type_counts = combined_contents.groupby(['platform','content_type']).size().reset_index(name = "count")
```

```python
content_type_counts
```

- 1.2 Group Content by language for each platform

```python
language_type_counts = combined_contents.groupby(['platform','language']).size().reset_index(name = "count")
```

```python
language_type_counts
```

- 1.3 Group content by Genre for each platform

```python
genre_type_counts = combined_contents.groupby(['platform','genre']).size().reset_index(name = "count")
```

```python
genre_type_counts
```

- 1.4 Group Content by Run Time for each platform

```python
run_time_counts = combined_contents.groupby(['platform','run_time']).size().reset_index(name = "count")
```
